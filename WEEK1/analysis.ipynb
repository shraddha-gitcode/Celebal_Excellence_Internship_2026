## 4. Handle Missing Values

```python
# Missing values - count per column
missing_count = df.isnull().sum()

# Missing values - percentage per column
missing_percent = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %': missing_percent.round(2)
})

# Show only columns that actually have missing values, worst first
missing_summary = missing_summary[missing_summary['Missing Count'] > 0]
missing_summary = missing_summary.sort_values('Missing Count', ascending=False)

print(missing_summary)
```

```python
# Total missing values in the entire dataset
total_missing = df.isnull().sum().sum()
print("Total missing values in dataset:", total_missing)
```

```python
# Sanity check for "suspiciously identical" mean/median in numeric columns
# (a common sign of earlier mean-imputation upstream)
numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    mean_val = df[col].mean()
    median_val = df[col].median()
    if abs(mean_val - median_val) < 1e-6 and not float(mean_val).is_integer():
        print(f"Possible pre-imputed column: '{col}' (mean = median = {mean_val})")
```

### Handle missing values (if any are found)

Strategy used:
- **Numeric columns** (e.g. rating, discount, ratings_count): fill missing
  values with the **median** (robust to outliers), not the mean.
- **Text / categorical columns** (e.g. seller_name, seller_information,
  videos, variations): fill missing values with `"Not Available"` instead of
  dropping rows, since dropping would lose otherwise-valid product records.

This code only acts on columns that actually have missing values, so it is
safe to run even if a column has none.

```python
categorical_cols = df.select_dtypes(include='object').columns

# Fill numeric missing values with the column median
for col in numeric_cols:
    n_missing = df[col].isnull().sum()
    if n_missing > 0:
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"Filled {n_missing} missing value(s) in '{col}' with median: {median_val}")

# Fill categorical missing values with a clear placeholder
for col in categorical_cols:
    n_missing = df[col].isnull().sum()
    if n_missing > 0:
        df[col] = df[col].fillna('Not Available')
        print(f"Filled {n_missing} missing value(s) in '{col}' with 'Not Available'")
```

```python
# Verify - should be 0 now
print("Total missing values after handling:", df.isnull().sum().sum())
```

## 8. Create a Derived Column

```python
# Check current datatype of final_price (it usually comes in as text, e.g. "₹3,995.00")
print(df['final_price'].dtype)
df['final_price'].head()
```

```python
# Clean final_price: remove currency symbol, commas, and quotes, then convert to numeric
df['final_price'] = (
    df['final_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.replace('"', '', regex=False)
    .str.strip()
)

df['final_price'] = pd.to_numeric(df['final_price'], errors='coerce')

# IMPORTANT: errors='coerce' silently turns anything that fails to parse into NaN.
# Always check for that immediately after - don't assume the conversion was clean.
new_nulls = df['final_price'].isnull().sum()
print("New missing values introduced by price cleaning:", new_nulls)
```

```python
# If cleaning introduced any new missing final_price values, fall back to initial_price
# (i.e. assume no effective discount for that row, rather than leaving it blank)
if df['final_price'].isnull().sum() > 0:
    df['final_price'] = df['final_price'].fillna(df['initial_price'])
    print("Filled remaining missing final_price values using initial_price as fallback.")
```

```python
# Derived column: actual discount amount in rupees
df['discount_amount'] = df['initial_price'] - df['final_price']

df[['initial_price', 'final_price', 'discount_amount']].head()
```

```python
# Data quality check: compare the dataset's own 'discount' % against the
# discount % we can calculate from initial_price and final_price.
# Large mismatches usually mean final_price wasn't captured correctly for that row.
df['computed_discount_pct'] = ((df['initial_price'] - df['final_price']) / df['initial_price']) * 100
df['discount_mismatch'] = (df['computed_discount_pct'] - df['discount']).abs()

mismatch_count = (df['discount_mismatch'] > 5).sum()  # more than 5 percentage points off
print(f"Rows where listed discount% and calculated discount% disagree by >5 points: {mismatch_count}")
df.sort_values('discount_mismatch', ascending=False)[
    ['title', 'initial_price', 'final_price', 'discount', 'computed_discount_pct', 'discount_mismatch']
].head(10)
```

## 12. Feature Engineering

### 12.1 Popularity Metric

A product's `rating` alone is misleading: a product with a perfect 5.0 rating from only 3 reviews is not more "popular" or more trustworthy than a product rated 4.4 from 1,000+ reviews. To combine **rating** and **ratings_count** into a single, fair popularity score, we use a **weighted (Bayesian-style) average** — the same logic used by IMDB and many review platforms:

$$\text{popularity\_score} = \frac{(\text{ratings\_count} \times \text{rating}) + (C \times m)}{\text{ratings\_count} + C}$$

where `m` is the dataset's average rating and `C` is the dataset's average ratings_count. Products with very few reviews get pulled toward the overall average `m` (so a lone 5-star review can't dominate); products with many reviews are dominated by their own actual rating.

```python
# Weighted popularity score (Bayesian average) - combines rating and ratings_count
C = df['ratings_count'].mean()   # average number of ratings across all products
m = df['rating'].mean()          # average rating across all products

df['popularity_score'] = (df['ratings_count'] * df['rating'] + C * m) / (df['ratings_count'] + C)

print(f"C (avg ratings_count) = {C:.2f}")
print(f"m (avg rating) = {m:.2f}")
df[['title', 'rating', 'ratings_count', 'popularity_score']].sort_values('popularity_score', ascending=False).head(10)
```

Notice how this differs from sorting by raw `rating` alone — products with a perfect 5.0 but only a handful of reviews (e.g. 4-7 ratings) no longer top the list, because the weighted score correctly treats a small sample size as less trustworthy.

### 12.2 Extracting Information from Complex Fields

Several columns store JSON-like text rather than plain values:
- **`sizes`** — a list of available size options per product, e.g. `[{"size":"M"},{"size":"L"}]`
- **`breadcrumbs`** — the category navigation path, e.g. `Accessories > Women > Backpacks > ...`

We extract a **size-option count** (a proxy for how many variants a product offers) and a clean **department** label (Women / Men / Girls / Boys / Unisex / Home & Other) from these fields, since the raw `category` column has 97 distinct values — too granular for department-level analysis.

```python
import json

# Number of size options available per product (parsed from the 'sizes' JSON field)
def count_sizes(value):
    if pd.isna(value) or value == 'Not Available':
        return 0
    try:
        return len(json.loads(value))
    except Exception:
        return 0

df['num_size_options'] = df['sizes'].apply(count_sizes)

# Top-level department, parsed from the 'breadcrumbs' JSON field
def get_department(value):
    try:
        names = [item['name'] for item in json.loads(value)]
        for n in names:
            if n in ['Women', 'Men', 'Girls', 'Boys', 'Unisex', 'Unisex Kids']:
                return n
        return 'Home & Other'
    except Exception:
        return 'Unknown'

df['department'] = df['breadcrumbs'].apply(get_department)

print("num_size_options summary:")
print(df['num_size_options'].describe())
print()
print("department counts:")
print(df['department'].value_counts())
```

## 16. Key Insights & Business Implications

**1. Listed discounts often don't match actual prices.**
After cleaning `final_price`, **36.3% of products (363 of 1,000)** have `final_price` exactly equal to `initial_price` — yet many of these same rows still list a nonzero `discount` percentage. Cross-checking the dataset's own `discount` column against a price-derived discount shows **423 rows** disagree by more than 5 percentage points. *Business implication:* discount badges/marketing claims should not be trusted at face value for analytics or compliance purposes; the actual transacted price should always be verified independently before reporting savings to customers or stakeholders.

**2. Raw rating is a misleading popularity signal.**
Sorting by raw `rating` surfaces products with perfect 5.0 scores from only 4-7 reviews, while products with thousands of reviews and a 4.4 rating rank lower. The weighted `popularity_score` corrects this. *Business implication:* any "Top Rated" or "Best Sellers" shelf should be built on a review-volume-aware score, not raw average rating, to avoid promoting under-reviewed or potentially fake-reviewed listings.

**3. Unreviewed products are being miscounted as low-rated.**
114 products show `rating = 0`, and in every case `ratings_count` is also 0 — these are new/unreviewed listings, not genuinely zero-star products. *Business implication:* filter these out (or label them "Not Yet Rated") before computing average ratings by category or seller, or the category's true quality will be understated.

**4. Category quality and discounting are unevenly distributed.**
`sports-shoes` and `sweaters` have the strongest average ratings (~4.1+) among top categories. `earrings` has both the *lowest* average rating (2.86) and the *highest* average discount (68.6%) of the top 10 categories. *Business implication:* heavy discounting in `earrings` is not translating into better customer satisfaction — this category may need a quality/sourcing review rather than continued price-based promotion.

**5. The catalog is heavily skewed toward Women's apparel.**
55% of all products belong to the Women's department, with `tops`, `dresses`, and `shirts` alone accounting for nearly a third of the entire catalog. *Business implication:* assortment and marketing strategy should reflect this concentration; categories with very thin representation (dozens of categories have just 1 product) are not yet meaningfully searchable or comparison-shoppable for customers and may need either expansion or removal from primary navigation.

**6. Size-option count varies widely by product.**
Products offer anywhere from 1 to 84 size/variant options (median 12). This metadata could be used as a proxy for catalog complexity or fulfillment risk (more SKU variants per listing generally means more inventory management overhead).
